# BFS

```python
def BFS(start,goal):
    to_do = [start]
    visited = []
    order = []

    while to_do:
      room = to_do.pop(0)

      if room in visited:
         continue
      visited.append(room)
      order.append(room)

      if room == goal:
          return order
      for nxt in map_rooms[room]:
          to_do.append(nxt)
    return order

  

order= BFS("A" ,"G")
print("BFS checked the rooms in this order:",order)
print("Numbers of rooms BFS checked:",len(order))
```

```python
def DFS(start,goal):
    to_do = [start]
    visited = []
    order = []

    while to_do:
      room = to_do.pop()

      if room in visited:
         continue
      visited.append(room)
      order.append(room)

      if room == goal:
          return order
      for nxt in map_rooms[room]:
          to_do.append(nxt)
    return order

  

order= DFS("A" ,"G")
print("DFS checked the rooms in this order:",order)
print("Numbers of rooms DFS checked:",len(order))
```
